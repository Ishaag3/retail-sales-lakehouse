**Dynamic Capabilities**

```python
dbutils.widgets.text("file_name","")
```

```python
p_file_name = dbutils.widgets.get("file_name")
```

### **Data Reading**

```python
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format","parquet")\
    .option("cloudFiles.schemaLocation",f"abfss://bronze@eteproj.dfs.core.windows.net/chheckpoint_{p_file_name}")\
    .load(f"abfss://source@eteproj.dfs.core.windows.net/{p_file_name}")
```

**Data Writing**

```python
df.writeStream.format("parquet")\
    .outputMode("append")\
    .option("checkpointLocation",f"abfss://bronze@eteproj.dfs.core.windows.net/{p_file_name}")\
    .option("path",f"abfss://bronze@eteproj.dfs.core.windows.net/{p_file_name}")\
    .trigger(availableNow=True)\
    .start()
```
